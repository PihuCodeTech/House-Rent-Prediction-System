# Baseline (mean rent) — Housing Rent Prediction

A naive constant predictor. Every real model must beat this. Uses the identical prepared data via `src/`.

```python
import sys, pathlib, warnings
sys.path.insert(0, str(pathlib.Path.cwd().parent))
warnings.filterwarnings('ignore')
from src.preprocessing import prepare_data, assert_run_consistency
from src.models import make_model, display_name
from src.evaluation import evaluate, results_table, print_summary, save_results
from src import config
```

```python
MODEL_KEY = 'baseline_mean'
EVALUATE_TEST = False
```

```python
data = prepare_data()
ref = assert_run_consistency(data)
n_rows = sum(data['meta']['rows'][s] for s in ('train','val','test'))
print(f"Prepared: {len(data['feature_names'])} features | reference_id {ref}")
```

```python
model = make_model(MODEL_KEY, data['target_transform'])
model
```

```python
results, pred, _ = evaluate(model, data, display_name(MODEL_KEY), MODEL_KEY, evaluate_test=EVALUATE_TEST)
print_summary(results, n_rows, len(data['feature_names']))
save_results(results, pred, data)
results_table(results)
```
